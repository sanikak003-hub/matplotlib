```python
import matplotlib.pyplot as plt
```

```python
day=[1,2,3,4,5,6,7]
no=[2,3,2,4,5,3,6]
plt.scatter (day,no,color="r",alpha=0.5,marker="*",edgecolor="g",linewidth=4)
t=plt.colorbar()
t.set_label("colorBar",fontsize=15)
plt.title("scatter plot",fontsize=15)
plt.xlabel("Day")
plt.ylabel("No")
```
